```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannotations.Annotator import Annotator
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.serif'] = ['Times New Roman']
```

```python
data = pd.read_csv(r"C:\Users\皮皮虾\Desktop\gene.csv",index_col=0)
fig = plt.figure(figsize = (15,10), dpi = 200)
color = ["#6388BB","#62B9A8","#A889BC"]
ax = fig.subplots()
sns.violinplot(data = data, ax = ax, x = 'group',
               y = 'expression', hue = 'group',
               palette = color, alpha = 0.7)
plt.tick_params(axis = 'both', labelsize = 20)
plt.xlabel(xlabel = 'group', fontsize = 20)
plt.ylabel(ylabel = 'expression', fontsize = 20)


sns.boxplot(data = data, ax = ax,x = 'group',y = 'expression',width = 0.1,color = 'white')
pairs = [
    ("Control", "Treatment1"),
    ("Control", "Treatment2"),
    ("Treatment1", "Treatment2")
]

annotator = Annotator(ax, pairs, data = data, x = "group", y = "expression")
annotator.configure(test="Mann-Whitney",
                    text_format="full",
                    loc="outside")
annotator.apply_and_annotate()
plt.savefig("violin_box_sig.png",dpi=300, bbox_inches="tight")
plt.show()
```
