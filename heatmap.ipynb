```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.serif'] = 'Times New Roman'
```

```python
data = pd.read_excel(r"C:\Users\皮皮虾\Desktop\data.xlsx", index_col=0)
data1 = data.values
```

```python
fig = plt.figure(figsize=(20, 10), dpi=200)
ax = fig.subplots()
my_color = ['#2c473e','#b8d9c9']
im = plt.imshow(data1, cmap = LinearSegmentedColormap.from_list('my_colormap',my_color), vmin = 0)
idx = data.index
col = data.columns
ax.set_xticks(ticks = np.arange(len(col)), labels = col, fontsize = 20)
ax.set_yticks(ticks = np.arange(len(idx)), labels = idx, fontsize = 20)
ax.set_xlabel('amino acid', fontsize = 20)
ax.set_ylabel('protein id', fontsize = 20)
cbar = plt.colorbar(im, ax = ax)
cbar.ax.tick_params(labelsize = 20)
cbar.ax.set_title('Ratio', fontsize = 20)
for i in range(len(idx)):
    for j in range(len(col)):
        plt.text(j, i, data1[i][j], fontsize = 20, ha = "center", va = "center", rotation = 45)
plt.savefig('D:/pythonproject/heatmap.png')
plt.show()
```
